```python
import os
import torch
from transformers import AutoProcessor, CLIPModel
from transformers.image_utils import load_image
```

```python
from huggingface_hub import login
token = input("get token info ?")
login(token=token)
```

```python
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
processor = AutoProcessor.from_pretrained("openai/clip-vit-base-patch32")
file_path = os.path.abspath(r'C:\Users\rahul\OneDrive\Desktop\data\images_test_IT_Roma00_00025_1\sample_set\sub_sample')

```

```python
label_groups = {'lighting':['day','night'],
                'vehicle': ['cars','bus'],
                'traffic': ['vehicle in same direction', 'vehicle in opposite direction']}

for files in os.listdir(file_path):
    image = load_image(os.path.join(file_path, files))
    for name, labels in label_groups.items():
        inputs = processor(text=labels, images=image, return_tensors="pt", padding=True)
        with torch.inference_mode():
            outputs = model(**inputs)
            logits_per_image = outputs.logits_per_image  # this is the image-text similarity score
            probs = logits_per_image.softmax(dim=1).numpy()[0]  # we can take the softmax to get the label probabilities
            best = probs.argmax().item()
            print(probs)
            print(labels[best])
            print(f"{probs[best]:.2f}")
    print("\n")
```
